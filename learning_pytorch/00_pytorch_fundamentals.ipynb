# PyTorch Fundamentals

```python
import torch
print(torch.__version__)
```

### Introduction to Tensors

```python
# scalar
scalar = torch.tensor(7)
scalar
```

```python
# vector
vector = torch.tensor([7, 7])
vector
```

```python
# matrix
matrix = torch.tensor([[7, 7], [7, 7]])
matrix
```

```python
# tensor
tensor = torch.tensor([[[1, 2, 3], [3, 6, 9], [2, 4, 5]]])
tensor
```

### Random Tensors

```python
# creating a random tensor of size (3, 4)
random_tensor = torch.rand(3, 4)
random_tensor
```

```python
# creating a random tensor with similar shape to an image tensor
random_image_size_tensor = torch.rand(size=(224, 224, 3)) # height, width, color channels (R, G, B)
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

### Zeros and ones

```python
# creating a tensor of all zeros
zeros = torch.zeros(size=(3, 4))
zeros
```

```python
# creating a tensor of all ones
ones = torch.ones(size=(3, 4))
ones
```

### Creating a range of tensors and tensors-like

```python
# use torch.range
one_to_ten = torch.arange(1, 11)
one_to_ten
```

```python
# creating tensors like
ten_zeros = torch.zeros_like(input=one_to_ten)
ten_zeros
```

### Tensor datatypes

**One of three problems in deep learning**
1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not on the right device

```python
# float 32 tensor
# the default is float32 even if you specify none
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype = None, # datatype is the tensor (float32, float16, etc..)
                               device = "None", # default is cpu, but can switch to cuda
                               requires_grad = False) # whether or not to track gradients with this tensors operations
float_32_tensor
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_16_tensor * float_32_tensor
```

```python
int_32_tensor = torch.tensor([3, 6, 9], dtype=torch.int32)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

**Getting information from tensors (tensor attributes)**

1. Tensors not right datatype - to get datatype from a tensor, can use `tensor.dtype`
2. Tensors not right shape - to get shape from a tensor, can use `tensor.shape`
3. Tensors not on the right device - to get device from a tensor, can use `tensor.device`

```python
# creating a random tensor
some_tensor = torch.rand(3, 4)
some_tensor
```

```python
print(some_tensor)
print(f"Datatype of tensor: {some_tensor.dtype}")
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Device tensor is on: {some_tensor.device}")
```
