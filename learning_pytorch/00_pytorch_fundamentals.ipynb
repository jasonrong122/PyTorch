### Tensor datatypes

**One of three problems in deep learning**
1. Tensors not right datatype
2. Tensors not right shape
3. Tensors not on the right device

```python
# float 32 tensor
# the default is float32 even if you specify none
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype = None, # datatype is the tensor (float32, float16, etc..)
                               device = "None", # default is cpu, but can switch to cuda
                               requires_grad = False) # whether or not to track gradients with this tensors operations
float_32_tensor
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_16_tensor * float_32_tensor
```

```python
int_32_tensor = torch.tensor([3, 6, 9], dtype=torch.int32)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

**Getting information from tensors (tensor attributes)**

1. Tensors not right datatype - to get datatype from a tensor, can use `tensor.dtype`
2. Tensors not right shape - to get shape from a tensor, can use `tensor.shape`
3. Tensors not on the right device - to get device from a tensor, can use `tensor.device`

```python
# creating a random tensor
some_tensor = torch.rand(3, 4)
some_tensor
```

```python
print(some_tensor)
print(f"Datatype of tensor: {some_tensor.dtype}")
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Device tensor is on: {some_tensor.device}")
```

### Matrix Multiplication

Two main ways of performing multiplication in neural networks and deep learning:
1. Element-wise multiplication
2. Matrix Multiplication (dot product)

There are two main rules that performing matrix multiplication needs to satisfy:
1. The **inner dimensions** must match:
* `(3, 2) @ (3, 2)` wont' work
* `(2, 3) @ (3, 2)` will work
* `(3, 2) @ (2, 3)` will work
2. The resulting matrix has the shape of the **outer dimensions**:
* `(2, 3) @ (3, 2)` -> `(2, 2)`
* `(3, 2) @ (2, 3)` -> `(3, 3)`

```python
# element-wise multiplication
tensor * tensor
```

```python
# matrix multiplication
torch.matmul(tensor, tensor)
```

### One of the most common errors in deep learning: shape errors

```python
# shapes for matrix multiplication
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])

tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])

# torch.mm(tensor_A, tensor_B) # torch.mm is the same as torch.matmul
# torch.matmaul(tensor_A, tensor_B) # this results in an error
torch.matmul(tensor_A, tensor_B.T)
```

```python
tensor_A.shape, tensor_B.shape
```

To fix our tensor shape issues, we can manipulate the shape of one of our tensors using **transpose**

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
```

```python
# the matrix multiplication operation works when tensor_B is trnasposed

torch.matmul(tensor_A, tensor_B.T)
```

### Finding the min, max, mean, sum, etc. (tensor aggregation)

```python
# creating a tensor
x = torch.arange(0, 100, 10)
x
```

```python
# find the min
torch.min(x), x.min()
```

```python
# find the max
torch.max(x), x.max()
```

```python
# find the mean - note: torch.mean() requires a tensor of float32 datatype to work
torch.mean(x.type(torch.float32)), x.type(torch.float32).mean()
```

```python
# find the sum
torch.sum(x), x.sum()
```
